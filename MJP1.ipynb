```python
import pandas as pd
df = pd.read_csv("/content/01_ClinTox_Molecular_Dataset.csv")

print("Original Shape:", df.shape)
print(df.head())
print(df.info())
print("\nMissing Values:")
print(df.isnull().sum())
df = df.dropna()
df = df.drop_duplicates()
required_columns = [
    "SMILES",
    "FDA_APPROVED",
    "CT_TOX"
]

df = df[required_columns]

print("\nFinal Shape:", df.shape)
df.to_csv("ClinTox_Preprocessed.csv", index=False)

print("\nClinTox Preprocessing Completed Successfully!")
```

```python
import pandas as pd
bac = pd.read_csv("/content/03_BacBench_Antibiotic_Resistance_Dataset.csv")
print("Original Shape:", bac.shape)
bac = bac.drop_duplicates()
bac = bac.dropna(subset=["genome_name"])
replace_dict = {
    "Sensitive": 0,
    "Resistant": 1,
    "S": 0,
    "R": 1
}

bac = bac.replace(replace_dict)

print("Final Shape:", bac.shape)

bac.to_csv("BacBench_Preprocessed.csv", index=False)

print("BacBench Preprocessing Completed Successfully!")
```
